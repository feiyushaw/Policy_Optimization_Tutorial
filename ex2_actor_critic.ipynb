# 练习内容

本练习基于 `2_policy_gradient.ipynb` 和 `3_actor_critic.ipynb` 的内容，通过实验进一步理解 Actor-Critic 方法。


## 练习 1：比较批量式与在线式 Actor-Critic

比较 `3_actor_critic.ipynb` 中介绍的两种 Actor-Critic：在线更新（online update）和批量更新（batch update）。至少选择两个环境，其中一个使用 `LunarLander-v2`。

为保证比较尽可能公平，两种方法应使用相同的策略网络结构。学习率可以沿用教程给出的设置。

强化学习的目标是最大化累计回报的期望，因此这里使用**平均累计回报**评估策略。注意这与 `ex1_evolutionary_policy_optimization.ipynb` 的评价方式不同。

每种方法使用 5 个不同随机种子独立训练，最大环境交互步数设为 $10^6$，绘制学习曲线。正式运行前建议先使用 $10^4$ 等较小步数检查代码和绘图逻辑。

`LunarLander-v2` 的每个 episode 长度不同，因此横轴应使用环境交互步数，而不是 episode 数。


## 练习 2：策略与价值函数表示能力的影响

对在线式 Actor-Critic，改变 Actor 和 Critic 的隐藏层宽度 `dim_hidden`，研究模型表示能力对学习效果的影响。

建议测试隐藏节点数：32、64、128、256。分别改变 Actor 和 Critic 的宽度，其余实验设置与练习 1 保持一致。

分析时至少关注：最终平均回报、学习速度、不同随机种子之间的方差，以及模型容量增大后是否出现收益饱和或训练不稳定。
