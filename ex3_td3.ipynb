# 练习内容

本练习围绕 `4_deterministic_policy_gradient.ipynb` 中介绍的 TD3（Twin Delayed Deep Deterministic Policy Gradient）展开。


## 练习 1：消融实验（Ablation Study）

针对教程中 TD3 引入的关键设计逐项做消融实验，验证每个组件的作用。完整枚举所有组件组合会带来很大的实验量，因此更常见的做法是：以完整 TD3 为基线，每次只移除一个组件，然后与完整方法比较。

建议检查的典型组件包括：

1. Twin Critics：使用两个 Q 网络并取较小值；
2. Delayed Policy Update：降低 Actor 的更新频率；
3. Target Policy Smoothing：在目标动作上加入裁剪噪声；
4. Target Network：使用慢更新的目标网络；
5. Replay Buffer：使用经验回放进行 off-policy 学习；
6. Exploration Noise：与环境交互时向动作加入探索噪声。

某个组件的效果未必能在单一环境中清楚体现。至少选择两个连续状态、连续动作环境，其中一个可使用 `Pendulum-v1`。

每个环境需要比较完整 TD3 与各个消融版本。绘图时应优先保证不同方法的学习速度、最终性能和稳定性能够直接比较。


## 练习 2：比较多种连续控制算法（选做）

可以使用 `Stable-Baselines3` 中的实现，将 TD3 与其他连续控制算法进行比较，例如 SAC、DDPG 等。

建议在多个环境、多个随机种子下比较平均累计回报、样本效率和训练稳定性，并结合各算法的设计差异解释结果。
